```python
import numpy as np

##################  数组 ##################
print("################## 数组 ##################")
l = np.array([1, 2, 3, 4, 5])
for idx, val in enumerate(l):
    val *= 2
    print(f"Index: {idx}, Value: {val}")
print(l)
print(l[0:3])


################## 元组 ##################
print("################## 元组 ##################")
t = (6, 7, 8, 9)
for idx, val in enumerate(t):
    print(f"Index: {idx}, Value: {val}")
print(t)


################## 字典 ##################
print("################## 字典 ##################")
d = {"name":"tom", "age":20}

# 默认遍历key
for k in d:
    print(k)

# 遍历 key+value（最常用）
for k, v in d.items():
    print(k, v)

# 只遍历value
for v in d.values():
    print(v)

# 只遍历key
for k in d.keys():
    print(k)
    
################## 集合 ##################
print("################## 集合 ##################")
s = {1, 2, 3, 4, 5}
for val in s:
    print(val)
```
