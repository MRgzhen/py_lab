```python
from ultralytics import YOLO

yolo = YOLO("icon.pt", task="detect")
# 图片
result = yolo(source="iconimg.png", save=True)
```

```python
display(result[0].names)
display(result[0].boxes)
display(result[0].boxes.cls)
display(result[0].boxes.cls)
```

```python
def get_icon_id(icon_name: str):
    for k, v in result[0].names.items():
        if v == icon_name:
            return k


def get_cls_no(icon_id: int):
    for k, v in enumerate(result[0].boxes.cls):
        if v == icon_id:
            return k


def get_xy(cls_no: int):
    x = result[0].boxes.xywh[cls_no][0]
    y = result[0].boxes.xywh[cls_no][1]
    return x, y

icon_id = get_icon_id("garbage")
if icon_id is None:
    raise Exception("no such icon")
cls_no = get_cls_no(icon_id)
display(cls_no)
pos = get_xy(cls_no)
display(pos)
```
